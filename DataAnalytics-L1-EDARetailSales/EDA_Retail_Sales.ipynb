```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Month'] = df['Date'].dt.to_period('M').astype(str)
```

```python
df['Quarter'] = df['Date'].dt.to_period('Q').astype(str)
```

```python
df[['Date', 'Month', 'Quarter']].head()
```

```python
monthly_sales = df.groupby('Month')['Total_Amount'].sum().reset_index()

monthly_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales['Month'], monthly_sales['Total_Amount'], marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

```python
quarterly_sales = df.groupby('Quarter')['Total_Amount'].sum().reset_index()

quarterly_sales
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    quarterly_sales['Quarter'],
    quarterly_sales['Total_Amount'],
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('Product_Category')['Total_Amount'].sum()
category_sales
```

```python
category_sales.sort_values().plot(
    kind='barh',
    figsize=(8, 5)
)

plt.title('Sales by Product Category')
plt.xlabel('Total Sales')
plt.ylabel('Product Category')
plt.tight_layout()
plt.show()
```
